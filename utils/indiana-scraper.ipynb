```python
for ov in glob("overview/*.json"):
    overview = json.load(open(ov))
    if overview["Results"] is None:
        continue

    finished = True
    for caseov in overview["Results"]:
        dstname = f"cases/{caseov['CaseNumber']}.json"
        if not os.path.exists(dstname):
            finished = False

    if finished:
        continue
    #proxy_index = (proxy_index +1 ) % len(proxy_list)
    #proxies=dict(https=proxy_list[proxy_index])
    proxies = {}
    req = {"Mode":"ByCase",
           "CaseNum":f"{ov.split("/")[-1][:-5]}*",
           "CiteNum":None,
           "CrossRefNum":None,
           "First":None,
           "Middle":None,
           "Last":None,
           "Business":None,"DoBStart":None,"DoBEnd":None,"OANum":None,"BarNum":None,"SoundEx":False,
           "CourtItemID":92,"Categories":None,"Limits":None,"Advanced":True,"ActiveFlag":"All","FileStart":None,
           "FileEnd":None,"CountyCode":None,"NewSearch":True,"CaptchaAnswer":None,"Skip":0,"Take":2000,"Sort":"FileDate DESC"}
    try:
        res = requests.post("https://public.courts.in.gov/mycase/Search/SearchCases", json=req, headers=hdr, proxies=proxies) #, verify=False
        overview = res.json()
    except:
        continue
        
    #print(ovd)
    print("use ", ov)
    for caseov in tqdm(overview["Results"]):
        dstname = f"cases/{caseov['CaseNumber']}.json"
        print("check", dstname)
        if not os.path.exists(dstname):
            try:
                #proxy_index = (proxy_index +1 ) % len(proxy_list)
                #proxies=dict(https=proxy_list[proxy_index])
                #print("use ", proxies)
                detail = requests.get(f"https://public.courts.in.gov/mycase/Case/CaseSummary?SRCT=&CaseToken={caseov["CaseToken"]}&_={int(datetime.now().timestamp())}",
                          headers=hdr, proxies=proxies) #, verify=False
                with open(dstname, "wb") as f:
                    f.write(detail.content)
            except Exception as e:
                print("could not query, skip", e.__repr__())
                pass
            time.sleep(5)
```
